Here is the order of increasing severity as defined by the python logging module. They are ***5** in number:

***DEBUG(10) --> INFO(20) --> WARNING(30) --> ERROR(40) --> CRITICAL(50)***

The values 10 to 50 represent the numeric values associated with the logging levels.

**1. DEBUG** <br>
Logs ***detailed information*** for debugging purposes and understanding the flow of the application during development. <br>
For example you can place different informative debug messages or intermediate results inside the source code in places like:
- Inside loop
- specific code blocks
- detailed variable values
- functional calls
- control information inside if else conditional statements to determine which conditions were met and which path the application took.

This is an alternative to the obvious print statements.
For some application, it may be neccessary to monitor most steps, hence the debug messages can be verbose.  <br>

***logging.debug("Execution at the first elif")***

 <br>

**2. INFO** <br>
Used to give general high-level information about the application's state or progress or significant event.
INFO logs are useful for monitoring the application's behaviour.
Here are important INFO logs:

- Startup messages
- Successful completion of major tasks or milestones which may include important variables.


***logging.info(f"Execution started")***

 <br>
 
**3. WARNING** <br>
Warnings indicate that something unexpected happened, or there may be a problem in the near future.
Alerts developers to situations that should be reviewed.

**4. ERROR** <br>
This level indicates a more serious problem that prevented a function from performing its intended task.

**5. CRITICAL** <br>
Critical messages indicate a grave error. 
Programs may not be able to continue running after encountering a critical error.

**Configuring the default logger**
You configure default logger only to log certain severity or higher, hence maintaining clean logs. This is done logging module's *basicConfig()*.<b>

The *basicConfig()* function takes the following kwargs among others:
(What is odd about the naming convention of basicConfiguration() function?)

- format: how the log statement is structured
- level: Debug severity
- filename: file name to log to
- filemode: mode to open file, default is append. You can specify others.

You will see log messages printed to the console with different severity levels based on the simulated actions. Adjust the logging level in basicConfig to see more or fewer log messages based on the desired severity threshold.

*NB:*
basicConfig() is only configured once in the life/instance of the program.
Note behaviour of running your program on notebook verses as a *.py script

Example:

```python
# Example configuring logging severity
import logging

# Configure logging
# logging.basicConfig(format='%(asctime)s - %(levelname)s - %(message)s', level=logging.DEBUG)

# This configuration is only done once in your program. 
# Example adding other parameters

logging.basicConfig(format='%(asctime)s - %(levelname)s - %(message)s',
                    datefmt='%d - %b - %y %H:%M:%S', # specify the time format
                    level=logging.DEBUG)

# Change the logging level to desired severity
logger = logging.getLogger()
logger.setLevel(logging.INFO)
print(logger, logger.level)
# logging.getLogger().setLevel(logging.INFO) # logger reconfiguration using a chained alternative
 

def example_logging_function():
    # DEBUG level
    logging.debug('This is a debug message.')

    # INFO level
    logging.info('This is an informational message.')

    # WARNING level
    logging.warning('This is a warning message.')

    # ERROR level
    logging.error('This is an error message.')

    # CRITICAL level
    logging.critical('This is a critical message.')

```
