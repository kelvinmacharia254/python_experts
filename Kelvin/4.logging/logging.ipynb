## LOGGING MODULE

It's a built-in module for logging errors, and it can also use for logging intermediate results as means to debug instead of using print statements.

The logging module is designed to log based on the severity of an event. 
Logging is basically setting up your program to info you when issues occur, say bugs among other runtime errors e.t.c.
Using logging effectively can help you address runtime issues that may arise which may not have been envisaged.

You often hear software engineers say 'lets look at the logs.'

Example log from PostgresSQL: 
Postgres Relational Database Management System *root* user is referred to as a *postgres* not root as in MySQL.

![postgres logs](postgres_logs.jpg)

Here is the order of increasing severity:
- DEBUG
- INFO
- WARNING
- ERROR
- CRITICAL

You configure default logger only to log certain severity or higher.
