## LOGGING MODULE

It's a built-in module for logging errors, and it can also use for logging intermediate results as means to debug instead of using print statements. 
Print statement can serve the purpose but for more structured and configurable approach, stick to the logs.

Logging is a perfect fit because it can be disabled with a single setting without commenting out or deletion of lines of unnecessary codes.

The logging module is designed to log based on the **severity** of an event. 
Logging is basically setting up your program to inform you when issues occur, say bugs among other runtime errors e.t.c.

Using logging effectively can help you address runtime issues that may arise which may not have been envisaged. 
It also reports critical activities to generally report how the application is fairing.

You often hear software engineers say 'lets look at the logs' when things go wrong or just to see how the application is fairing to plan for possible improvements.
You have also noted, in case of a crash or issue with some application you are request to submit logs to the developers.Most desktop applications come with a feature that allows submission of logs. 
Logs are nothing much, but than technical details of how an application is fairing, it may contain time, dates, type of errors or even inputs that triggered failure in case it occurred.
You can also log tracebacks.

Here is the order of increasing severity as defined by the python logging module. They are ***5** in number:

***DEBUG(10) --> INFO(20) --> WARNING(30) --> ERROR(40) --> CRITICAL(50)***

The values 10 to 50 represent the numeric values associated with the logging levels.

**1. DEBUG** <br>
Logs ***detailed information*** for debugging purposes and understanding the flow of the application during development. <br>
For example, you can place different informative debug messages or intermediate results inside the source code in places like:
- Inside loop
- specific code blocks
- detailed variable values
- functional calls
- control information inside if else conditional statements to determine which conditions were met and which path the application took.

This is an alternative to the obvious print statements.
For some applications, it may be necessary to monitor most steps; hence the debug messages can be verbose.  <br>

***logging.debug("Execution at the first elif")***

 <br>

**2. INFO** <br>
Used to give general high-level information about the application's state or progress or significant event.
INFO logs are useful for monitoring the application's behaviour.
Here are important INFO logs:

- Startup messages
- Successful completion of major tasks or milestones which may include important variables.


***logging.info(f"Execution started")***

 <br>
 
**3. WARNING** <br>
Warnings indicate that something unexpected happened, or there may be a problem in the near future.
Alerts developers to situations that should be reviewed.

**4. ERROR** <br>
This level indicates a more serious problem that prevented a function from performing its intended task.

**5. CRITICAL** <br>
Critical messages indicate a grave error. 
Programs may not be able to continue running after encountering a critical error.

**NOTE:**
The default level is WARNING, only events of this level and above will be tracked and logged, unless the logging package
is configured to do otherwise using the basicConfig() or setLevel() function.

**Configuring the default logger**
You configure default logger only to log certain severity or higher, hence maintaining clean logs. This is done logging module's *basicConfig()*.<b>

The *basicConfig()* function takes the following kwargs among others:
(What is odd about the naming convention of basicConfiguration() function?)

- format: how the log statement is structured
- level: Debug severity
- filename: file name to log to
- filemode: mode to open file, default is append. You can specify others.

You will see log messages printed to the console with different severity levels based on the simulated actions. Adjust the logging level in basicConfig to see more or fewer log messages based on the desired severity threshold.

**NOTE:**
basicConfig() is only configured once in the life/instance of the program.

### Logging tracebacks using logging module
Python by default sends log messages to the standard error(stderr).

To ensure our code doesn't brake, we use try and except block to enclose possible areas in our code that may cause errors. 
try and except blocks handle errors gracefully. Once exceptions are handled, we log them to provide vital information for debugging.

Why log tracebacks?
1. Problem identification so that a fix can be provided. We can store logs for latter analysis unlike when they are displayed on stdout/console.

2. Silence or hide exception by logging. As a good developer, you think of scenarios but that may not be possible most times. Handle errors gracefully to avoid issues in production as you figure them out in future fixes and patches.
 Logging helps you identify error occurring, and you fix and patch accordingly.
