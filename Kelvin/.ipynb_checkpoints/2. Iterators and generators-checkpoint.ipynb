## Iterables and Iterators 

Iteration is running a piece of code several times until a condition or a task is accomplished.

Repetitive task in coding are common, you can hardly solve many coding problems without them. For this reason apart for **while** and **for** loops python provides additional tools.

Python even provides [itertools](https://docs.python.org/3/library/itertools.html) inbuilt module to help with task involving repetitions.
This module saves, implementing some algos from first principle.
The  itertool modules comes with functions such as **permutations**, **combinations**, **product** and  etal.


### Iterables

Iterables: This are objects that can be iterated/looped over using either while or for loops.
Iterables includes list, tuples, strings, files, generators etc.

For an object to be a iterable, it must have a special method **`__iter__`** as part of its
functionality.

To check methods and attributes available to an object, the inbuilt **dir** function. 

```python
# Iterables example
# Iterating through a list using a for loop
names =['Carol','Rogers','Kelvin'] # iterable
for name in names:
    print(name)

# use dir to inspects the names list
print(dir(names)) # 
# The names list has an __iter__ method and hence it's an iterable. The __iter__ is what 
# ..runs behind the scenes to enable looping. This concept is clearer if you have done some OOP basics.
```

```python
# Iterating through a list using a while loop
names =['Carol','Rogers','Kelvin'] # iterable
count = 0
while count < len(names):
    print(names[count])
    count +=1
```

### Iterator
An Iterator is an object with a state such that it remembers where it is during iteration.

An iterator is still an iterable but with an additional functionality of state which is impelemented by **`__next__`**.

```python
# Create an Iterator from an iterable using __iter__() method. The relationship between the two is circular.
names =['Carol','Rogers','Kelvin'] # iterable
i_names = names.__iter__() # or iter(names)
# inspect the iterator
print(dir(i_names))

print("****************************************")
# it can be looped over just like an iterable
for name in i_names:
    print(name)
```

```python
# unique functionality of state. Call next() and pass the iterator as an argument
names =['Carol','Rogers','Kelvin'] # iterable
i_names = names.__iter__() # or iter(names)

# unique functionality of state. Call next() and pass the iterator as an argument
print(next(i_names))
print(next(i_names))
print(next(i_names))
```

### Application of Iterables and Iterators

To solves our coding problems, we write our own functions and classes. With classes, we can make the iterable to suite our requirements.

```python
class Portfolio:
    def __init__(self):
        self.holdings = {}  # key = company, value = number of shares

    def buy(self, company, shares):
        self.holdings[company] = self.holdings.get(company, 0) + shares

    def sell(self, company, shares):
        if company in self.holdings:
            current_shares = self.holdings[company]
            if shares > current_shares:
                print(f"Error: Not enough shares of {company} to sell.")
            else:
                new_shares = current_shares - shares
                if new_shares == 0:
                    del self.holdings[company]  # Remove entry if shares become zero
                else:
                    self.holdings[company] = new_shares
        else:
            print(f"Error: Company '{company}' not found in the portfolio.")

    def __iter__(self):
        return iter(self.holdings.items())

print("*** Customer_1 Portfolio **")
customer_1 = Portfolio()
customer_1.buy('KPLC', 1000)
customer_1.buy('CENTUM', 205)
customer_1.buy('KQ', 110)
customer_1.buy('KQ', 360)

print(customer_1.holdings)
for (company, shares) in customer_1:
    print(company, shares)

print("*** Customer_2 Portfolio **")
customer_2 = Portfolio()
customer_2.buy('KCB', 10000)
customer_2.buy('Alphabet', 2050)
customer_2.buy('Kakuzi', 110000)

print(customer_2.holdings)
for (company, shares) in customer_2:
    print(company, shares)
```

**The essence of creating an iterable class is to enable convenient iteration over its elements. By implementing __iter__ and returning an iterator (in this case, the items of the holdings dictionary), the class allows for cleaner and more readable iteration, making it easier to work with the contents of the portfolio.**
