### Virtual Environments
Python has vast library of both inbuilt and 3rd party modules in the Python Package Index(PyPI). PyPI is a central repo host python packages.
To install a package into your PC, the Python Package Index(PIP).
> pip install <package_name>

It is against good software development practices to install all packages into the global space of your PC. Different projects you work on require different packages/dependencies.

The concept of virtual environment was created to isolate each projects whereby it can install its own dependencies without 'polluting' the global python environment.

Most common used packages for virtual environment management are 'venv' and 'pipenv'
1. venv
   venv ships with python.
   To create a virtual environment,  create a project directory and run
   > python -m venv <venv_name> # devs like using .venv as the venv_name

   To activate a virtual env run,
   > .\venv_name\Scripts\activate # windows
   or
   > source /venv_name/bin/activate # linux, wsl, macOS

2. pipenv
    Pipenv is a popular open-source dependency and virtual environment management tool for python projects developed by Python Packaging Authority.
    To use it, you have to install it from PyPI using PIP
    > pip install pipenv

    General configuration of virtual envs  with `pipenv` to manage your project's dependencies.

    a. **Create or Navigate to Your Project Directory:**
       Make sure you are in the directory where your Python project is located or where you want to create a new project.

    b. **Initialize a `pipenv` Environment:**
       If you haven't already initialized a `pipenv` environment for your project, you can do so by running:
       2.
       > pipenv --python 3.x  # Replace 3.x with your desired Python version

       This command will create a new `Pipfile` and a virtual environment.

    c.  **Install Dependencies from `requirements.txt` or individually:**
        - If you have a `requirements.txt` file that lists your project's dependencies, you can install them into your `pipenv` virtual environment using the following command:

       > pipenv install -r requirements.txt

       This command will read the dependencies from the `requirements.txt` file and install them in your virtual environment.

       To istall individual packages, the pipenv venv must be activated.
       If the virtual environment is not already activated, you can activate it with:

       > pipenv shell

       Once activated, you'll be in the virtual environment, and you can work on your project with the installed dependencies.

    d. **Deactivate the Virtual Environment (When Done):**
       To deactivate the virtual environment and return to your system's default Python environment, you can use:

       > exit

       or

       > deactivate



# RUNNING PYTHON SCRIPTS

**Prerequisites:** Working with Command Line Interface(CLI) on OSs

We write code to run it to solve a problem. 

Depending on requirements of your project, python provides vast way to execute scripts. 
For instance, I am using jupyter right now to do this presentation.

A python interpreter converts code to binary for computer understand.

Python code can be in the form of:
1. Scripts: Simple scripts ran directly
2. Modules/packages: meant to be imported
3. Executable (complete compiled projects .exe file)

Here are some of the ways to run python code.
1. Interactive/Read Evaluate Print Loop(REPL): Demo with IDLE and CLI
2. Scripts from windows/linus/macOS : Double-click on a file_name.py
   This scripts can be directly or as imports and packages
3. Running from Terminal/CMD. 
   You have an option to pass arguments if there are required but completely options. 
   This is configured using sys.argv or 3rdy party package argparse.
   use commands:  
   ../../> python script.py arg1 arg2 .. 
   ../../> python script.py arg1 arg2 .. 
4. Packages: Live demo and give example of packages on the PyPI e.g. Django, Numpy e.t.c

**Note:**
1. Add Python to PATH: To run Python scripts without specifying the full path to the Python interpreter, 
   you can ensure that Python is added to your system's PATH environment variable during installation. 
   Many Python installers for Windows provide an option to do this automatically. Read more on [Add python to environment variable in windows.](https://realpython.com/add-python-to-path/).
2. In Unix-based operating systems like Ubuntu or Mac use **python3**. (you can set aliases though)
