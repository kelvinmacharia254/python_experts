# Dictionary Review

A dictionary is a built-in Python data structure that stores data in key-value pairs.

```python
# Example:
person = {
    "name": "Kelvin",
    "location": "Nairobi",
    "language": "Python"
}

person
```

Used be unorderd before version 3.7

```python
# retrieval when key is known
person["name"]
```

```python
# retrieval when key is not known
# use method to avoid KeyError
# person["sport"] # this gives an error
value = person.get("name", "The key doesn't exit.")  # default is None
print(value)
```

```python
# updating
person["name"] = "Caroline"
value = person.get("name", "The key doesn't exit.")  # default is None
print(person)
print(value)
```

```python
# fetching keys by 
for item in person:
    print(item)
```

```python
# fetching value by looping
for item in person:
    print(f"{item}: {person[item]}")
```

```python
# inspect a dic attrs using dir()
dir(person)
# help(person.keys)
```

```python
# getting keys usinf keys() method
for k in person.keys():
    print(k)
```

```python
# getting values using values() method
for v in person.values():
    print(v)
```

```python
# retriew each key value as a tuple
for item in person.items():
    print(item)
```

```python
for item in person.items():
    print(item[0], item[1])
```

```python
for k, v in person.items():
    print(k, v)
```

```python
#  illustration how to unpack a tuple
t = (1,2,3)
first,  second, third = t
print(first)
print(second)
print(third)
t[0] = 8
```

```python
# count letters in a word
s = "banana"
count = dict()
for letter in s:
    count[letter] = count.get(letter, 0) + 1

print(count)
```
