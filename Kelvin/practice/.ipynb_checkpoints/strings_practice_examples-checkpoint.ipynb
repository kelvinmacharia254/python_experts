```python
"""
Title Case a Sentence
Write a function that converts a sentence into title case (the first letter of each word capitalized).
Example Input: "this is a test sentence"
Example Output: "This Is A Test Sentence"
"""

def convert_title_case(s):
    return s.title()
s = "this is a test sentence"
print(convert_title_case(s))
print("\n********************")
help(s.title)
```

```python
"""
Count Substring Occurrences
Write a function that counts how many times a substring appears in a given string (case-sensitive).
Example Input: "ababababab", "aba"
Example Output: 2
"""
s, substring = "ababababab" , "aba" # lowercase
s1, substring1 = "caronjokinjokicaronjoki" , "Njoki" # uppercase
def count_substring(s, substring):
    return s.count(substring)

print(count_substring(s, substring))
print(count_substring(s1, substring1))
print("\n********************")
help(s.count)
```

```python
""""
Replace All Digits
Write a function that replaces all digits (0-9) in a string with a given character.
Example Input: "P1y2t3h4o5n", "#"
Example Output: "P#y#t#h#o#n"
"""
s, replacement = "P1y2t3h4o5n", "#"
def replace_digits(s, replacement):
    result = ""
    for letter in s:
        if letter.isdigit():
            result += replacement
        else:
            result += letter
    return result

print(replace_digits(s, replacement))
print("\n********************")
help(s.isdigit)


# alternate code using list comprehension
def replace_digits(s, replacement):
    return ''.join(replacement if ch.isdigit() else ch for ch in s)
```

```python
"""
Check if String Starts or Ends with Substring
Write a function that checks if a given string starts with a specified prefix and ends with a specified suffix (case-sensitive).
Example Input: "Hello, World!", "Hello", "World!"
Example Output: True
"""
s, pre, suf = "Hello, World!", "Hello", "World!"
def check_string(s, prefix, suffix):
    return all([s.startswith(prefix), s.endswith(suffix)])

print(check_string(s, pre, suf ))
print("\n********************")
print(help(all))
print(help(s.startswith))
print("\n********************")
print(help(s.endswith))
```

```python
# """
# Encode a String (Caesar Cipher)
# Write a function that shifts each letter in a string by a given number (Caesar cipher). 
# Only letters should be shifted; numbers and symbols should remain unchanged.
# Example Input: "abc XYZ", 3
# Example Output: "def ABC"
# """
import string
letters =  string.ascii_letters # abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ
print(letters)
letters_count = len(letters)

def caeser_cipher(s, offset):
    results = "" 
    for letter in s:
        if letter in letters:
            index = letters.index(letter)
            n = index + offset
            if index < int(letters_count/2) :
                if n > (int(letters_count/2) - 1):
                    results += letters[(n - (int(letters_count/2)))]
                else:
                    results += letters[n]
            else:
                if n > (letters_count - 1):
                    results += letters[(n - (int(letters_count/2)))]
                else:
                    results += letters[n] 
        else:
            results += letter
            
    return results                

# Test Case 0: Mixed cases
print("***********************************")
print("Test Case 0")
s, offset = "abc XYZ", 3 # results: 'def ABC'
print(caeser_cipher(s, offset))

# Test Case 1: Basic shift, lowercase
print("***********************************")
print("Test Case 1")
s = "hello"
offset = 1
print(caeser_cipher(s, offset))
# Expected output: "ifmmp"

# Test Case 2: Basic shift, uppercase
print("***********************************")
print("Test Case 2")
s = "PYTHON"
offset = 2
print(caeser_cipher(s, offset))
# Expected output: "RAVJQP"

# Test Case 3: Wrapping around (lowercase)
print("***********************************")
print("Test Case 3")
s = "zebra"
offset = 3
print(caeser_cipher(s, offset))
# Expected output: "cheud"

# Test Case 4: Wrapping around (uppercase)
print("***********************************")
print("Test Case 4")
s = "APPLE"
offset = 25 # Shift by -1, so P becomes O, etc.
print(caeser_cipher(s, offset))
# Expected output: "ZOOKD"

# Test Case 5: Zero offset (no change)
print("***********************************")
print("Test Case 5")
s = "Programming"
offset = 0
print(caeser_cipher(s, offset))
# Expected output: "Programming"

# Test Case 6: Large offset (multiple wraps)
print("***********************************")
print("Test Case 6")
s = "coding"
offset = 28 # Equivalent to offset 2 (28 % 26 = 2)
print(caeser_cipher(s, offset))
# Expected output: "eqfkpi"

# Test Case 7: String with spaces and punctuation
print("***********************************")
print("Test Case 7")
s = "Hello, World!"
offset = 5
print(caeser_cipher(s, offset))
# Expected output: "Mjqqt, Bwzqi!"

# Test Case 8: String with numbers and symbols
print("***********************************")
print("Test Case 8")
s = "123 Go!"
offset = 4
print(caeser_cipher(s, offset))
# Expected output: "123 Ks!"

# Test Case 9: Empty string
print("***********************************")
print("Test Case 9")
s = ""
offset = 10
print(caeser_cipher(s, offset))
# Expected output: ""

# Test Case 10: String with only non-alphabetic characters
print("***********************************")
print("Test Case 10")
s = "!@#$%^&*"
offset = 7
print(caeser_cipher(s, offset))
# Expected output: "!@#$%^&*"

# Test Case 11: Single letter, wrapping
print("***********************************")
print("Test Case 11")
s = "z"
offset = 1
print(caeser_cipher(s, offset))
# Expected output: "a"

# Test Case 12: Single letter, large offset
print("***********************************")
print("Test Case 12")
s = "A"
offset = 53 # Equivalent to offset 1 (53 % 26 = 1)
print(caeser_cipher(s, offset))
# Expected output: "B"
```
