We encounter errors when running python scripts.
Python has an inbuilt way to let us know when things go wrong, but it can be challenging to understand without any knowledge of what these error messages mean.

Traceback is what you see on the console when your code fails. 

This text will explain the details provided in the tracebacks to help debug. Understanding traceback can save a lot of time used in debugging.

## Types of Errors:
### 1. Syntax Errors
These are about the structure and grammar of the programming language, preventing it from even starting to run.
Syntax errors are easier to catch because they prevent code execution.
### 2. Runtime Errors
These errors occur while the program is running, hence the term "runtime."
Runtime errors are caused by issues that occur while the program is being executed. 
It could be due to unforeseen user input, a missing file, dividing by zero, or attempting to access an item beyond the bounds of a list.
Handling runtime errors often involves thorough testing, error-checking, and anticipating various scenarios that could lead to issues during execution.

Example 1: Simple traceback

```python
def addTwoNumbers(x, y):

    return x + y

print(addTwoNumbers(1, q))
```

Example 2: Chained traceback

```python
def addTwoNumbers2(x, y):
    z = x+q
    return z

print(addTwoNumbers2(1, 2))
```

- Traceback Chain/block is read from the bottom to top(bottom up). 
- At the bottom before execution stops, you will find the name of the exception that occurred with some explanation, e.g. ***"NameError: name 'q' is not defined"*** as in the example above.
- The recent line of code to execute and caused the exception then follows immediately above.
- The traceback provides a chronological order of how the code ran from the beginning until the error occurred but must be read in reverse. 
- This is intuitive because reading from the bottom to top you are finding out the trace, kinda forensic audit.
- Tracing back, you are led to where it all has begun, i.e., the first trigger or the first call to initiate the code execution but resulted in an exception.
- From here, you can even tell what really triggered the errors. 
- If it is a syntactical error, then solve by following the rules of the language.
- If it is a runtime, the real bounty hunt follows. 

The example above is just a simple function and a function call statement. 
In complex programs like desktop apps, personal projects or frameworks such as Django, the source code contains a lot of variables, classes, files, functions, statements among other python objects. When errors occur, the traceback can be a bunch but with the skills of reading through and interpreting tracebacks can aid debugging to provide error handling mechanism.

### COMMON TRACEBACK ERRORS

#### 1. AttributeError
   Occurs when you try to use an object attribute that doesn't exist.

```python
# Example AttributeError
# sort() is a list Attribute but sorted is not
lst = [9,7,6,4,5,3,1,8,2]
lst.sort() # sorts the list in place
lst
```

```python
lst = [9,7,6,4,5,3,1,8,2]
lst.sorted() # AttributeError
```

Sorted() is a python builtin python function. It applies to most iterables such as list, tuples and strings.
It doesn't edit the iterable in place but rather makes a copy of it.

```python
# Example using sorted function
name = 'minnesota'
sorted(name) # make a copy
name # no change will be observed on the original string
```

```python
# Example using sorted function
name = 'minnesota'
sorted_name = sorted(name)
print(sorted_name)
sorted_name = ''.join(sorted_name) # join to display sorted as string
print(sorted_name)
```

#### 3. IndexError
   Occurs when you retrieve an index from an iterable from a list that doesn't exist.

```python
lst = [9,7,6,4,5,3,1,8,2]
print(f"The list has 9 elements with index from 0 to {len(lst)-1}")
for i, m in enumerate(lst):
    print(f"Element{i+1} --> index = {i} --> element = {m}")
lst[15] 
```

#### 3. KeyError
   Similar in nature with IndexError but raised by python dictionaries when you access 
   a key that doesn't exist.

```python
python_experts ={
    'django_guy':'Kelvin',
    'data_scientist': 'Kemmie',
    'data_analytics': 'Rogers'
}
print("***using for loop to access a dictionary***")
for name, value in python_experts.items():
    print(f"{name}: {value}")

print("***Accessing by keys***")
print(python_experts['django_guy'])
print(python_experts['data_scientist'])
print(python_experts['data_analytics'])

print("***Try a dictionary keys that is non-existence***")
print(python_experts['GUI_guy'])
```
