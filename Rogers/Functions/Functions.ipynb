### Gazillion Lines, the Old-Fashioned Way

You have been tasked with developing a very important calculator function that adds two numbers and prints out the sum. How would we go about it in the old-fashioned way?

```python
# We need to develop a program that adds two numbers, the old-fashioned way
num_1 = 2
num_2 = 3
sum = num_1 + num_2
# we expect the sum to be 5, or is it?
print(sum)
```

We have provided 4 lines of code. What if we want to repeat the calculation three times?

```python
# round 1
num_1 = 2
num_2 = 3
sum = num_1 + num_2
print(sum)

#round 2
num_1 = 4
num_2 = 3
sum = num_1 + num_2
print(sum)

#round 3
num_1 = 4
num_2 = 5
sum = num_1 + num_2
print(sum)
```

Here, we have 12 lines of code to execute the summation of three pairs of numbers. Imagine now if we need to develop a more elaborate NASA-level calculator that has multiplication, division, addition and subtraction. We would be dealing with maybe a gazillion lines of code, as we will need a new set of code for every new execution of the mathematical functions.

Huh! You did not see it coming, did you? Functions. Functions in mathematics are a set of "rules" that allow for solving of problems within a sub-domain. An example is the equation of a straight line, y = mx+c. This function applies to all straight lines. You do not have to derive the properties of a straight line from first principles every time you encounter one.

What if we transfer the concept of functions from mathematics to our Sum program?

### Functions, the New Way

```python
# defining a function
def sum_num(x,y):
    '''
    creating a 'template' that can be applied to any two sets of numbers that require summing.
    '''
    sum = x + y
    print(sum)

# performing calculations to three sets of numbers
sum_num(2,3)
sum_num(3,4)
sum_num(4,5)
```

We have the same output as the old-fashioned code, but with much less lines of code: 6 (50% less code).

## What are functions in programming?

- named code block that performs some actions with the purpose of computing a final value or result, which is then sent back to the caller code.

It is important to differentiate it from a **procedure**, which is a named code block that performs a set of actions without computing a final value or result.

## Why do we write functions?
- Allows for easier implementation of a program map, as functions allow us to conceive our program as a bunch of sub-steps
- allow for re-use of code
- allows for re-use of variable names, therefore manageable code namespace
- allows for isolated testing of sections of our program
- allows for change or fixing of code in only one location

### 3. Function Arguments

#### *Positional Arguments/Parameters*

In the example function **definition** `def sum_num(x,y):`, the x & y are referred to as **positonal arguments** or **required arguments**. They are a comma-separated list within the function parenthesis. The arguments can also be referred to as **local variables**



```python
# function that computes sum of two numbers and outputs in the form x+y=sum
def sum_num(x,y):
    sum = x + y
    return print(x,"+",y,"=",sum)
```

When the function is **called**, a corresponding list of arguments MUST be provided.
1. `sum_num()` will raise a TypeError as the instance is missing 2 required positional arguments, as shown below:

```python
sum_num()
```

2. `sum_num(2,3,4)` will raise a TypeError as the instance is providing more arguments than is required

```python
sum_num(2,3,4)
```

3. `sum_num(2,3)`. value of `2` is passed to `x` and value of `3` is passed to `y`. Therefore within the instance of the function, `x=2` and `y=3`.

```python
## instance 1 of the sum_num function
sum_num(2,3)
```

4.  `sum_num(4,5)`, the function's local variables will have values as follows: `x=4` and `y=5`.

```python
## instance 2 of the sum_num function
sum_num(4,5)
```

You will note that the values provided in the function instance are passed to the function's local variable in sequential order.

#### *Keyword Arguments*

When you’re calling a function, you can specify arguments in the form `<`keyword`>`=`<`value`>`. In that case, each `<`keyword`>` must match a parameter in the Python function definition.

```python
sum_num(x=4, y=5)
```

```python
sum_num(y=5, x=4)
```

You can now notice that the order of the arguments does not matter. The output is exactly the same. Now, what if instead of positional parameter `y`, we use `z`?

```python
sum_num(x=4, z=5)
```

A TypeError is raised, as `z` is not one of the declared positional parameters

One characteristic is common, though, between positional arguments and keyword arguments: number of arguments and parameters must match.

```python
sum_num(x=4)
```

```python
sum_num(x=4, y=5, z=6)
```

Also, you can mix positional and keyword styles of passing arguments, as shown, BUT all the positional arguments must come first:

```python
# positional argument is first
sum_num(4, y=5)
```

```python
# keyword argument is first
sum_num(x=4, 5)
```

#### *Default Parameter*

Default parameters are specified when defining a python function, and take the form of `<`name`>`=`<`value`>`, where `<`value`>` is the default value for that parameter. 

Default parameters are **defined ony once** when the function is defined (executed ie), and therefore the default value is not redefined each time the function is called.

```python
def sales(item, price, max_discount, min_price=200):
    if price < min_price:
        price = min_price
    if max_discount > 0.10:
        max_discount = 10
    print("The price of", item, "is ", price, "and a maximum discount of", max_discount, "% is aplicable.")
```

In this example, the value of 200 for `min_price` is the default value for that parameter. If the sale() is called, any argument that's left out assumes its default value:

```python
sales('Banana', 150, 0.15) 
# min_price argument has been left out. 
#note tha the price argument is less than the default value of min_price
```

CAUTION!!

Do not specify a default parameter value that is a mutable object eg. list. This is best explained through an example:

```python
def list_append(my_list = []): # append passed values and "###" string to a list
    my_list.append('###')
    return (my_list)
```

```python
# First run of the called function
print(list_append(['foo', 'bar', 'baz']))
print(list_append([1, 2, 3, 4, 5]))
```

```python
list_append()
```

Now, what if we run the function without any parameters twice?

```python
list_append()
```

```python
list_append()
```

```python
print(list_append([1, 2, 3, 4, 5]))
```

Now, the function is not behaving as we would instinctly expect. Though we can see that the mis-behaviour is not present when arguments are passed.

It is best practice to pass `None` as default value of a parameter, as follows:

```python
def list_append(my_list = None): # append passed values and "###" string to a list
    if my_list is None:
        my_list = []
    my_list.append('###')
    return (my_list)
```

So now, even if we run the function five times, the output wil be predictably a list of length = 1.

```python
list_append()
list_append()
list_append()
list_append()
list_append()
```

### 4. `return`ing Function Value

Function value is the final value of computations handled within a function.
An important part of a function is the builtin `return` keyword. It enables the return of the function value to the caller code.

We can rewrite `sum_num()` to incorporate the `return` keyword:

```python
# defining a function
def sum_num(x,y): # function that computes the sum of two numbers
    '''
    creating a 'template' that can be applied to any two sets of numbers that require summing.
    eg. sum of 1 + 2 = 3
    return value is 3
    '''
    return (x + y)

# performing calculations to three sets of numbers
print(sum_num(2,3))
print(sum_num(3,4))
print(sum_num(4,5))
```

Our code is now 5 lines. Note that we have passed the summing operation directly to the return keyword. This helps to reduce the lines of code. Only ensure that it is clear.

#### *Explicit Return Statement*

Functions basically can return any Python object to the caller code, using an explicit return statement:
- numeric values `int, float, and complex` values, 
- collections and sequences of objects `list, tuple, dictionary, or set` objects, 
- binary values `True` or `False`
- boolean operator `not`
- comparison operators `==, !=, >, >=, <, <=`
- membership operator `in`
- identity operator `is`
- user-defined objects, 
- classes, 
- functions,
- modules or packages
- ...

Note that you can return more than one object, eg, `return sum, x, y`

#### *Implicit Return Statement*

An inplicit return statement (return is omitted, or a bare return without a return value is used) returns the following:
- `None`

```python
# with a return statement
def sum_num(x,y): # function that computes the sum of two numbers
    sum = x + y
    return sum
    
value = sum_num(2,3) # expect 5 as sum
print (value)
```

```python
# With no return statement
def sum_num(x,y): # function that computes the sum of two numbers
    sum = x + y
    
value = sum_num(2,3) # expect 5 as sum
print (value)
```

Implicit return statements can be used to generate *side effects* such as:
- printing something to the screen, expecially useful when scripting. An example is our first implementation of sum_num()
- modifying a global variable (NOT RECOMMENDED)
- updating the state of an object
- writing some text to a file
- ...

```python
# describing how to change a global variable
counter = 0
def increment(): # a function that changes a global variable
    global counter
    counter += 1

increment()
print(counter)
```

The example above shows a function that changes a global variable. The result of calling `increment()` will depend on the initial value of `counter`. Different initial values for `counter` will generate different results, so the function’s result can’t be controlled by the function itself.

Modifying global variables is generally considered a bad programming practice. Just like programs with complex expressions, programs that modify global variables can be **difficult to debug, understand, and maintain**. The function uses the `global` statement, which is also considered a bad programming practice in Python.

When you modify a global variable, you’re potentially affecting all the functions, classes, objects, and any other parts of your programs that rely on that global variable.

To understand a program that modifies global variables, you need to be aware of all the parts of the program that can see, access, and change those variables. So, **good practice recommends writing self-contained functions that take some arguments and return a useful value (or values) without causing any side effect on global variables**.

A remedy would be to write a self-contained `increment()` that takes arguments and returns a coherent value that **depends only on the input arguments**, as shown below:

```python
counter = 0

def increment(var):
    return var + 1

print(increment (counter))
```

```python
print (counter) ## value of counter is not changed outside of increment() 
                 # therefore does not lead to unexpected issues in the program
```

When there is need to change the value of `counter`, it will be done so explicitly with a call to `increment()`

```python
# explicit increase to the value of counter by 1
counter = increment(counter)
print (counter)
```

This way, you’ll have more control over what’s happening with `counter` throughout your code.

##### *Using returns with Conditions*

You can actually have several return statements within a function, BUT, this is only if they are within condtional statements. An example is:

```python
type(2)
```

```python
def sum_num(x,y):
    if type(x) is int and type(y) is int: # ensuring argument values are integers
        return x+y
    else: # what to do if argument values are not integers
        return "Error. One or all values are not integers"

print(sum_num(2,3))
print(sum_num('a',2))
print(sum_num('a','b'))
```

Otherwise, if have two return statements within a function, the function will terminate after encountering the first returnstatement. Therefore, it wil not evaluate the other return statements (also known as **dead code**), as shown below:

```python
def sum_num(x,y):
    return x+y
    return "Pass. All values are integers" # dead code

print(sum_num(2,3))
```

In general, a function takes **arguments** (in our case: x & y), **performs** some operations, and **returns** a value (or object). The value that a function returns to the caller is generally known as the function’s return value.
