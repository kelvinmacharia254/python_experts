### Programming paradigms

<span style="color:grey">

"Programming paradigms are different ways or styles in which a given program or programming language can be organized."

</span>
by FreeCodeCamp

Classification of programming langauges into paradigms is intended to guide users on best languages to use for certain scenarios and types of problems. It is worthy to note that a programming language can be described using several paradigms, and that the **paradigms are NOT mutually exclusive**, even though some have been built to satisfy requirements of a specific paradigm.

The paradigms are described below:

#### 1. Imperative:

consists of sets of detailed instructions that are given to the computer to execute in a given order. Imperative programming focuses on describing how a program operates, step by step, from top to bottom.

Illustration: procedure for baking a cake
```
1- Pour flour in a bowl
2- Pour a couple eggs in the same bowl
3- Pour some milk in the same bowl
4- Mix the ingredients
5- Pour the mix in a mold
6- Cook for 35 minutes
7- Let chill
```

#### 2. Procedural: 

Follows a sequence of statements or commands in order to achieve a desired output. Each series of steps is called a procedure or subroutine (functions).

Example: C and C++, Java, Pascal, BASIC

Illustration: procedure for baking a cake
```
function pourIngredients() {
    - Pour flour in a bowl
    - Pour a couple eggs in the same bowl
    - Pour some milk in the same bowl
}

function mixAndTransferToMold() {
    - Mix the ingredients
    - Pour the mix in a mold
}

function cookAndLetChill() {
    - Cook for 35 minutes
    - Let chill
}

pourIngredients()
mixAndTransferToMold()
cookAndLetChill()
```

#### 3. Functional: 

Focus on the output of mathematical functions and evaluations. Each function–a reusable module of code–performs a specific task and returns a result. Functions can be assigned to variables, passed as arguments, and returned from other functions.

Examples: Scala, Erlang, Haskell, Elixir, F#

Illustration: 

Imperative programming
```
nums = [1,4,3,6,7,8,9,2]
result = [] # External variable
num_length = len(nums)

for i in num_length:
    if nums[i] > 5:
        result.append(i)

print(result) // Output: [ 6, 7, 8, 9 ]
```
Functional programming
```
nums = [1,4,3,6,7,8,9,2]

def filterNums():
    result = [] // Internal variable
    for i in num_length:
        if nums[i] > 5:
            result.append(i)
    return result

print(result) // Output: [ 6, 7, 8, 9 ]
```
When external variables are used to store a function's result, this can be considered as being a side effect. Functional programming ensures that a function does not modify any variable outside of its scope.

#### 4. Declarative: 

Characterised by statements which declare precisely the changes that need to be made. It is all about hiding away complexity and bringing programming languages closer to human language and thinking.

`filter`, `map`, `reduce` and `sort` functions are good examples of declarative code.

Examples: RDBMS eg. SQL

#### 5. Object-Oriented: 

Treats a program as a group of objects composed of data and program elements, known as attributes and methods. Objects can be reused within a program or in other programs. In OOP, classes are heavily used, and are used as blueprint (parent) and their properties are inherited by instances (child).

Examples: Java, Python, PHP, C++, Ruby

## CLASS METHODS VERSUS STATIC METHODS

Instance methods automatically take the instanc as the first argument `self`.

A class method, on the other hand, is created by adding a `@classmethod` decorator.

While instance methods and class methods pass the instance (`self`) and class (`cls`) automatically as the first argument, respectively, static methods do not pass anything automatically.

Static methods are used where an action is related to the class logically but is not dependent on any specific instance or class variable. Static methods are created by adding `@staticmethod` decorator before a regular method.

### Instance methods

````python
class Employee:
    
    raise_amount = 1.0 # implementing soft-coding, instead of  hardcoding the raise-amount value
    
    # defining class variables
    def __init__(self, firstName: str, lastName: str, salary: int, level: str):
        self.firstName = firstName
        self.lastName = lastName
        self.salary = salary
        self.level = level
        self.userEmail = self.firstName+'.'+self.lastName+'@pye.com' 
        
    def email(self) -> str:
        emailAddress = self.firstName+'.'+self.lastName+'@pye.com'
        return print(emailAddress)
    
    def apply_raise(self): # regular method
        self.salary = int(self.salary * self.raise_amount)
````

### Class methods

````python

class Employee:
    
    raise_amount = 1.0 # implementing soft-coding, instead of  hardcoding the raise-amount value
    
    # defining class variables
    def __init__(self, firstName: str, lastName: str, salary: int, level: str):
        self.firstName = firstName
        self.lastName = lastName
        self.salary = salary
        self.level = level
        self.userEmail = self.firstName+'.'+self.lastName+'@pye.com' 
        
    def email(self) -> str:
        emailAddress = self.firstName+'.'+self.lastName+'@pye.com'
        return print(emailAddress)
    
    
    def apply_raise(self): # regular method
        self.salary = int(self.salary * self.raise_amount)
    
    @classmethod # adding this decorator to convert regular method to class method
    def set_raise_rate(cls, rate): # cls used instead of self to rep. the class
        cls.raise_amount = rate
        
    def __str__(self):
        return self.firstName + " " + self.lastName
````

### Class Methods as Constructors

Convention: names begin with `from_`. eg. from_string, and are terminated by a `return` statement.

````python
# Example of Constructor that parses a string and creates an instance
# from a comma-delimited text containing  employee details

class Employee:
    
    raise_amount = 1.0 # implementing soft-coding, instead of  hardcoding the raise-amount value
    
    # defining class variables
    def __init__(self, firstName: str, lastName: str, salary: int, level: str):
        self.firstName = firstName
        self.lastName = lastName
        self.salary = salary
        self.level = level
        self.userEmail = self.firstName+'.'+self.lastName+'@pye.com' 
        
    def email(self) -> str:
        emailAddress = self.firstName+'.'+self.lastName+'@pye.com'
        return print(emailAddress)
    
    @classmethod # adding this decorator to convert regular method to class method
    def from_string(cls, string): # cls used instead of self to rep. the class
        first, last, salary, level =  string.split(',')
        return cls(first, last, salary, level) # creates and returns an instance. 
                                                # Similar to Employee(first, last, salary, level)
````

````python
# sample input

string = "Rogers,Mugambi,50000,'V'"
Bdev_4 = Employee.from_string(string)
print(Bdev_4.salary)
````

## INHERITANCE, POLYMORPHISM & ABSTRACTION

### Inheritance

Allows one class to acquire the properties (attributes) and behaviors (methods) of another class. 

The class being inherited from is called the parent class or superclass.

The class that inherits is called the child class or subclass.

````python
class Animal:  # Parent class
    def __init__(self, name):
        self.name = name
    def printName(self):
        print(self.name)

class Dog(Animal):  # Child class
    def bark(self):
        return "Woof!"

dog1 = Dog("Alfa")
dog1.printName()
print(dog1.bark())
````

Suppose we have a general Vehicle class that defines properties common to all vehicles, such as make, model, and year. Now, we want to create a Car class that has all the properties of Vehicle, but also some properties specific to cars, like number_of_doors.

````python
class Vehicle: # parent class
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

# within vehicles, we have several categories: cars, SUVs, Trucks, Motorcycles, Lorry

# class Car(Vehicle): # child class
#     def __init__(self, number_of_doors):
#         self.number_of_doors = number_of_doors
        
class Car(Vehicle): # child class
    def __init__(self, make, model, year, number_of_doors):
        super().__init__(make, model, year)
        self.number_of_doors = number_of_doors

car1 = Car("Toyota",'Corolla','2024',4)

class Lorry(Vehicle): # child class
    def __init__(self,make, model, year, axles):
        super().__init__(make, model, year)
        self.axles = axles
        
lorry1 = Lorry('Benz','Actros550','2024','10')
````

In this example, Car is a subclass of Vehicle and inherits its properties and methods. The super() function is used to call the __init__() method of the Vehicle superclass, which allows us to use its properties in the Car subclass.

### Subclasses

A subclass is a class that inherits from one or more classes. 

Python supports multiple inheritance, which means a subclass can have more than one superclass and inherit features from all superclasses.

Let’s extend our vehicle example. Suppose we want to create a Convertible class that inherits properties from both the Car and a ConvertibleFeatures class:

````python
class Vehicle: # parent class - Level 1
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

class Car(Vehicle): # child class - Level 2
    def __init__(self, make, model, year, number_of_doors):
        super().__init__(make, model, year)
        self.number_of_doors = number_of_doors

class ConvertibleFeatures: # parent class - Level 1
    def __init__(self, roof_type):
        self.roof_type = roof_type

class Convertible(Car, ConvertibleFeatures): # Inheritance from more than one class
    def __init__(self, make, model, year, number_of_doors, roof_type):
        Car.__init__(self, make, model, year, number_of_doors)
        ConvertibleFeatures.__init__(self, roof_type)
````

In this example, Convertible is a subclass that inherits from both the Car and ConvertibleFeatures classes. It uses the __init__() methods of both superclasses to inherit their properties.

While inheritance and subclasses can make our code more modular and easier to maintain, they can also make it more complex.

It is important to ensure the features with care (judiciously) and keep code as simple as possible.

### Polymorphism

Polymorphism allows us to use a single interface with different underlying forms. 

Polymorphism allows us to define methods in the child class with the same name as defined in their parent class.

````python
class Animal:  # Parent class
    def __init__(self, name):
        self.name = name

class Dog(Animal):  # Child class
    def bark(self):
        return "Woof!"

class Cat(Animal):  # Another child class of Animal
    def sound(self):
        return "Meow!"

# Using Polymorphism
def animal_sound(animal):
    print(animal.sound())

cat = Cat("Tom")
dog = Dog("Spike")

animal_sound(cat)  # Outputs: Meow!
animal_sound(dog)  # Outputs: Woof!
````

In this example, both Cat and Dog classes have a method named sound, but the implementations are different. This is polymorphism.

### Abstraction

Abstraction allows us to hide the internal implementation details and show only the functionality to the users. 

In Python, we achieve abstraction using abstract classes and interfaces.

````python
from abc import ABC, abstractmethod

class AbstractAnimal(ABC):  # Abstract Class
    @abstractmethod
    def sound(self):
        pass

class Dog(AbstractAnimal):  # Child class
    def sound(self):
        return "Woof!"
````

In this example, AbstractAnimal is an abstract class that declares the sound method but doesn’t provide any implementation. The Dog class provides the implementation for the sound method.

Example: 

consider a drawing application that supports shapes like Circle, Rectangle, and Triangle. Each of these shapes has a draw() method. When the user selects a shape and drags the mouse, the application doesn’t need to know what kind of shape it is - it just calls the draw() method. This is polymorphism. If you want to add a new shape to the application, you just create a new class for that shape and implement the draw() method. The existing code doesn’t need to change, which makes the system easy to extend.

#### Abstraction

Used to hide complexity of the system and expose only the necessary details.

Example: 

consider a car. A car is a complex machine with many parts like an engine, gearbox, etc. But when you drive a car, you don’t need to know how all these parts work - you only need to know how to use the steering wheel, accelerator, and brakes. This is abstraction. In programming, we achieve abstraction using interfaces and abstract classes. They allow us to hide the internal workings of a class and show only the methods and properties that the user needs to know about.

Example 2: 

let’s say you want to use the Python statistics module, which is a built-in module in Python. This means that Python already comes with the module in its library. You don’t need to import it with PIP2. To use this module, you just need to know how to call its functions. You don’t need to understand how these functions are implemented. This is an example of abstraction2.
