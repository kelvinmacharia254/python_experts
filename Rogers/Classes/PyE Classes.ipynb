## INTRODUCTION

### Programming paradigms

<span style="color:grey">

"Programming paradigms are different ways or styles in which a given program or programming language can be organized."

</span>
by FreeCodeCamp

Classification of programming langauges into paradigms is intended to guide users on best languages to use for certain scenarios and types of problems. It is worthy to note that a programming language can be described using several paradigms, and that the **paradigms are NOT mutually exclusive**, even though some have been built to satisfy requirements of a specific paradigm.

The paradigms are described below:

#### 1. Imperative:

consists of sets of detailed instructions that are given to the computer to execute in a given order. Imperative programming focuses on describing how a program operates, step by step, from top to bottom.

Illustration: procedure for baking a cake
```
1- Pour flour in a bowl
2- Pour a couple eggs in the same bowl
3- Pour some milk in the same bowl
4- Mix the ingredients
5- Pour the mix in a mold
6- Cook for 35 minutes
7- Let chill
```

#### 2. Procedural: 

Follows a sequence of statements or commands in order to achieve a desired output. Each series of steps is called a procedure or subroutine (functions).

Example: C and C++, Java, Pascal, BASIC

Illustration: procedure for baking a cake
```
function pourIngredients() {
    - Pour flour in a bowl
    - Pour a couple eggs in the same bowl
    - Pour some milk in the same bowl
}

function mixAndTransferToMold() {
    - Mix the ingredients
    - Pour the mix in a mold
}

function cookAndLetChill() {
    - Cook for 35 minutes
    - Let chill
}

pourIngredients()
mixAndTransferToMold()
cookAndLetChill()
```

#### 3. Functional: 

Focus on the output of mathematical functions and evaluations. Each function–a reusable module of code–performs a specific task and returns a result. Functions can be assigned to variables, passed as arguments, and returned from other functions.

Examples: Scala, Erlang, Haskell, Elixir, F#

Illustration: 

Imperative programming
```
nums = [1,4,3,6,7,8,9,2]
result = [] // External variable
num_length = len(nums)

for i in num_length:
    if nums[i] > 5:
        result.append(i)

print(result) // Output: [ 6, 7, 8, 9 ]
```
Functional programming
```
nums = [1,4,3,6,7,8,9,2]

def filterNums():
    result = [] // Internal variable
    for i in num_length:
        if nums[i] > 5:
            result.append(i)
    return result

print(result) // Output: [ 6, 7, 8, 9 ]
```
When external variables are used to store a function's result, this can be considered as being a side effect. Functional programming ensures that a function does not modify any variable outside of its scope.

#### 4. Declarative: 

Characterised by statements which declare precisely the changes that need to be made. It is all about hiding away complexity and bringing programming languages closer to human language and thinking.

`filter`, `map`, `reduce` and `sort` functions are good examples of declarative code.

Examples: RDBMS eg. SQL

#### 5. Object-Oriented: 

Treats a program as a group of objects composed of data and program elements, known as attributes and methods. Objects can be reused within a program or in other programs. In OOP, classes are heavily used, and are used as blueprint (parent) and their properties are inherited by instances (child).

Examples: Java, Python, PHP, C++, Ruby

Illustration:
```
// Create the two classes corresponding to each entity
class Cook {
	constructor constructor (name) {
        this.name = name
    }

    mixAndBake() {
        - Mix the ingredients
    	- Pour the mix in a mold
        - Cook for 35 minutes
    }
}

class AssistantCook {
    constructor (name) {
        this.name = name
    }

    pourIngredients() {
        - Pour flour in a bowl
        - Pour a couple eggs in the same bowl
        - Pour some milk in the same bowl
    }
    
    chillTheCake() {
    	- Let chill
    }
}

// Instantiate an object from each class
const Frank = new Cook('Frank')
const Anthony = new AssistantCook('Anthony')

// Call the corresponding methods from each instance
Anthony.pourIngredients()
Frank.mixAndBake()
Anthony.chillTheCake()
```

## PYTHON CLASS TERMINOLOGIES

While handling classes, we will come across several concepts and terminologies. These are as follows:

<img src="./Classes example.png">

* Attributes: variables defined inside a class with the purpose of storing all the required data for the class to work. Refer to the properties or data associated with a specific object of a given class.
* Methods: functions defined in the class. Refer to the different behaviors that objects will show
* Instance: a class object that has been derived from the class. Each instance will access all attributes and methods of the Parent class and can edit them without effecting changes in the parent class. New attributes and methods can also be introduced in the instances, which the parent class cannot access.
