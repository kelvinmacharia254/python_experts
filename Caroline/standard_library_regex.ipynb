# Standrad Libraries in python
- Standard libraries in python is a collection of built in modules and packages that comes bundled with python.
- These tools provide ready to use tools and functionalities, so you don't need to install anything extra to use them.
- They support common programming tasks such as file handling, regular expressions, math operations, data serialization etc.
- You just need to import them them to use them e.g import math, import re etc
- they save developmental time and promotes best practices.

### Examples of standard Libraries
| Module  | Description |
|---|---|
| Math| provides mathematical functions like pi, sqrt()|
| datetime | Works with dates and times | 
|OS | Interacts with operating system like file paths, directories etc |
| re | regular expressions matching and text searching |
| collections | specialized data structures like counter, deque, defaultdict |
| Statistics | performs statistical calculations like mean, median, and mode
| itertools | tools for efficient looping and combinations
|json | encode and decode json data
| time | time related fuctions like delays and timestamps
| sys | Access system specific parameters and functions
|random | generate random numbers and choices
| http | tools for working with HTTP(like http.client, http.server)

### What are Third Party libraries
- Third- party libraries in python are external packages or modules that are not included int he standard Python distribution.
- They are created by the python community and other developers to extend the Python capabilities especially in specialized fields like data science, web development, machine learning, game development and more.
- they are not built in and you need to install them separately(usually using pip)
- they are usually more powerful or specialized than standard libraries
- They are usually available from the Python Package Index(PyPI)



#### Example using Third-party libraries
- `pip install numpy as np` : goes ahead to install numpy and its dependecies
-  Numpy is used for numerical computing. much faster than import math which is a standard Library. processes arrays faster. Used for linear algebra and statistics. Random number generation etc. has good memory usage.
- `pip install library_name`
