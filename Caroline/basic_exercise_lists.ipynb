- **3-4**. Guest List: If you could invite anyone, living or deceased, to dinner, who would

Then use your list to print a message to each person, inviting them to dinner.

#### 3-4 Solution

```python
#inviting friends to dinner and printing out a message to each
friends = ['Biliah','Charlene','Joy']

print(f"Hello {friends[0]}, I'm inviting for dinner at my house tomorrow")
print(f"Hello {friends[1]}, I'm inviting for dinner at my house tomorrow")
print(f"Hello {friends[-1]}, I'm inviting for dinner at my house tomorrow")
```
