# Python zip()
- **Zip()** function takes iterables(can be zero or more), aggregates them in a tuple and returns it.
- syntax of zip()
```
zip(*iterables)
```

- *iterables* can be built-in iterables (like: list, string, dict) or user defined iterables


### Examples

````python
#aggregate languages and respective versions
languages = ['Java', 'Python', 'JavaScript']
versions = [14, 3, 6]

result = zip(languages, versions)

print(list(result))
````

````python
# example with number_list, str_list
number_list = [1, 2, 3]
str_list = ['one', 'two', 'three']

# No iterables are passed
result = zip()

# Converting iterator to list
result_list = list(result)
print(result_list)

# Two iterables are passed
result = zip(number_list, str_list)

# Converting iterator to list
#result_list = list(result)
#print(result_list)

# Converting iterator to set
result_set = set(result)
print(result_set)
````
