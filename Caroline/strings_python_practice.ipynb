#### 1. Reverse a string
- Write a function that takes a string and returns it reversed


```python
#write a function that takes a string and returns it reversed

#define a function for reversing a string
def reverse_st(str1):
    result = str1[::-1] #use slicing s atring for reversal
    return result


#define a main(to ask for user input, call function and print results)
def main():
    #prompt user to enter an input
    st = input(" Enter a string: ")
    #call function with user input
    reversed_string = reverse_st(st)
    #show result to the user
    print(f"Reversed string: '{reversed_string}'")

#main() runs the program
main()




```

### 2. Count Vowels: Write a function that counts vowel in a string(case sensitive)

```python
#count vowels in a string

def count_vowels(str2):
    #initialize count
    count = 0 

    #use 'in' operator to check if char in string
    for char in str2:
        if char in "aAeEiIoOuU":
            #increament count
            count += 1
    
    return count

def main():
    #prompt user to enter string
    text = input("Enter a string: ")

    #call function
    vowel_count = count_vowels(text)
    print(f"Number of vowels is: {vowel_count}")

main()
```

### 3.Remove Whitespace: Write a function to remove white spaces from a string 

```python
def remove_spaces(s):
    result = s.replace(" ", "") ##use replace() method
    return result

def main():
    text1 = input("Enter a string")

    #call function
    no_space_text = remove_spaces(text1)
    
    print(f"No spaces text: {no_space_text}")

main()
```

### 4.Find the Longest Word: Write a function that takes a sentence and returns the longest word in it. If there are multiple words of the same length, return the first one.


```python
def longest_word(text2):
    word = text2.split()
    longest_word = max(word, key = len)
    return longest_word

def main():
    sentence = input("Enter a sentence: ")
    longest = longest_word(sentence)
    print(f"Longest word in the sentence: {longest}")

main()
```

### Improve the code to be case_sensitive 
- Make use of Tenary operator


```python
def str_palindrome(str_to_check):
    #make input uppercase and remove spaces fro a better match
    str_input = str_to_check.replace(" ", "").upper()
    return "A palindrome" if str_input == str_input[::-1] else "Not a Palindrome"

def main():
    txt_to_check = input("Enter a string: ")
    String_palindrome = str_palindrome(txt_to_check)
    print(String_palindrome)
    

main()
```
