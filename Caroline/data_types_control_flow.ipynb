# Basic String Operations

### Concatenation
adding two strings using the  '+' operator



```python
first_name = "John"
last_name = "Doe"
full_name = first_name + " " + last_name
print(full_name) 
```

### Changing case in strings with methods
using *title()*, it Capitalizes the first letters in a word.
- A *method* is an action that python can perform on a piece of data

```python
name = "ada lovelace"
print(name.title())
```

```python
#other methods can be used to change cases
name = "Ada Lovelace"
print(name.upper())
print(name.lower())
```

### String Formatting (Using variables in strings)
- String formatting allows you to insert variables or expressions into strings

#### Using f-strings(formatted literal strings)

```python
first_name = "Ada"
last_name = "Lovelace"
full_name = f"my name is {first_name} {last_name}"
print(full_name)
```

```python
#add methods to the f-string commposed message
first_name = "ada"
last_name = "lovelace"
full_name = f"{first_name} {last_name}"
print(f" Hello, {full_name.title()}!")
```

```python
#assign f-string composed message to a variable
first_name = "ada"
last_name = "lovelace"
full_name = f"{first_name} {last_name}"
message = (f" Hello, {full_name.title()}!")
print(message)
```

### using the **'.format()'** method

### Adding whitespace to strings with tabs or newlines
- to add tab to your text, use the caracter combination \t
- to add newline to your text, use the caracter combination \n
- you can combine tab and newline in a single line \n\t


### Stripping whitespace
- python can look for extra whitespace on the left or right side of the string.
- **rstrip()**- whitespace exists on the right side of string
- **lstrip()**- whitespace exists on the left side of string
- **strip()**- whitespace exists on the either side of string

```python
favorite_language = "python "
favorite_language = favorite_language.rstrip()
favorite_language
```

### Other common string methods
- *'str.split()* : splits a strings into a list of substrings
- * 'str.join()'* : joins a list of strings into one string
- *str.replace*: replace a substring with another substring


```python
#example using str.split()
text = "Python is fun"
words = text.split()
print(words)
```

```python
#example using str.join
words = ['Python', 'is','fun']
text = ' '.join(words)
print(text)
```

```python
#example using str.replace
text = "Hello, world!"
new_text = text.replace("World", "Python")
print(new_text)
```

### Escape Characters
- Some characters have special meanings in strings and need to be escaped with a backslash (\) if you want to include them in a string:

```python
escaped_text = " This is a \"quoted\" string."
print(escaped_text)
```

# strings operations are immutable
- Strings in Python are immutable, meaning you cannot change individual characters in a string. You can create a new string with the desired modifications.

```python
#example of creating a new string with desired modification
text = "Hello, World!" 
new_text = text[:5] + ' ' + "Python!"
print(new_text)
```

### String Length
- You can find the length of a string using the len() function:

```python
# find string length
name =  "Caroline"
length = len(name)
print(length)
```

### String Repetition
- You can repeat a string using the * operator:

```python
name = "Caroline" * 4
print(name)
```

### Accessing Characters
- Python strings are zero-indexed, meaning you can access individual characters by their index:

```python
name = "Caroline"
print(name[0])
print(name[7])
```

# String Slicing

# SETS
- A set is an unordered collections of unique elements, duplicates are not allowed.
- sets are used to store distinct values
-A set itself may be modified but the elements contained in the set must be hashable.
- sets are defiend by enclosing a coma-separated sequence of elements with curly brcaes '{}' or by using built in '**set()**' cinstructor. 
[find link to resource](https://realpython.com/lessons/what-is-a-set/)

### creating sets

```python
# using curly braces
my_set1 = {2,3}

# Using the set() constructor
my_set2 = set([1,2,4,4])

print(my_set1, my_set2)
```

### Uniqueness:

- Sets do not allow duplicate elements. If you try to add a duplicate element, it will not be stored in the set:

```python
# duplicates are removed in sets
my_set = {2,3,4,5,5,4}
print(my_set)
```

### No Indexing:

- Sets are unordered, and elements are not indexed. Therefore, you cannot access set elements by their position as you can with lists or tuples

### Common Set Operations:

- Sets support a variety of set operations, including union, intersection, difference, and more:

- **Union ('|')**: Combines two sets, returning elements that are in either set.
- **Intersection (&)**: Returns elements that are common to both sets.
- **Difference (-)** : Returns elements that are in the first set but not in the second.
- **Symmetric Difference (^)** : Returns elements that are in either of the sets, but not both.
