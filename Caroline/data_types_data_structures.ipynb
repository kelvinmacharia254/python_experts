### String manipulation: Regular Expressions 

 In Python, regular expressions are implemented as part of the re module, which provides functions for working with regex patterns. Regular expressions are used to search, match, and manipulate strings based on patterns that you define. They allow you to:

Search for specific patterns in text.
Validate data based on a pattern (e.g., checking if an email address is in the correct format).
Extract specific information from text (e.g., extracting phone numbers from a document).
Replace or modify text based on patterns.
Here are some common concepts and components of regular expressions in Python:

Pattern Matching: Regular expressions are used to define patterns that specify what you're searching for in a string. These patterns can include specific characters, character classes, and quantifiers that dictate how many times a character or pattern should appear.

Metacharacters: Special characters with special meanings in regular expressions, such as . (matches any character except a newline), * (matches 0 or more occurrences), + (matches 1 or more occurrences), ? (matches 0 or 1 occurrence), and more.

Character Classes: A character class is a set of characters enclosed in square brackets, such as [aeiou], which matches any vowel. You can use character classes to specify a range of characters you want to match.

Quantifiers: Quantifiers dictate how many times a character or pattern should appear. For example, * means "zero or more occurrences," + means "one or more occurrences," and {n, m} means "between n and m occurrences."

Anchors: Anchors are special characters used to match positions within the text, such as ^ (start of a line), $ (end of a line), and \b (word boundary).

Grouping: You can use parentheses to group characters or patterns together. This is useful for applying quantifiers or other operations to a group of characters.

Modifiers: Flags like re.IGNORECASE (case-insensitive matching) and re.MULTILINE (multiline matching) allow you to modify how the regular expression pattern is applied.


```python
import re

text = "Please contact support@example.com or info@domain.co for assistance with your inquiry."

# Define a regex pattern to match email addresses
email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,7}\b'

# Find all email addresses in the text
email_addresses = re.findall(email_pattern, text, re.IGNORECASE)

# Print the extracted email addresses
for email in email_addresses:
    print(email)
```

**Explanation of Code**
- import re: Import the re module, which provides functions for working with regular expressions.

- text: A sample text that contains email addresses. You can replace this with your own text data.

- email_pattern: This is the regular expression pattern used to match email addresses. Let's break down the pattern piece by piece:

- \b: This is a word boundary anchor. It ensures that the email address is not part of a larger word. For example, it prevents matching "myemail@example.computer."

[A-Za-z0-9._%+-]+: This part matches the username part of an email address. Here's what it does:

[A-Za-z0-9]: Matches any alphanumeric character (both uppercase and lowercase) or a digit.
._%+-: Matches special characters often found in usernames, such as periods, underscores, percent signs, plus signs, and hyphens. These special characters are allowed in usernames.
@: Matches the "@" symbol, which separates the username from the domain.

[A-Za-z0-9.-]+: This part matches the domain part of an email address. Here's what it does:

[A-Za-z0-9]: Matches any alphanumeric character (both uppercase and lowercase) or a digit.
.-: Matches the period (dot) and hyphen, which are allowed characters in domain names.
\.: Matches the period (dot) that separates the domain from the top-level domain (TLD).

[A-Z|a-z]{2,7}: This part matches the top-level domain (TLD), ensuring that it consists of 2 to 7 characters. It uses character classes [A-Z|a-z] to match both uppercase and lowercase letters. This allows for TLDs like "com," "co," "edu," and more.

\b: Another word boundary anchor to ensure that the matched email address is a complete word and not part of a larger word.

re.IGNORECASE: The third argument in re.findall() is an optional flag. In this case, re.IGNORECASE is used to perform a case-insensitive search, allowing the regex to match email addresses regardless of case (e.g., "Support@example.com" or "support@Example.com").

re.findall(email_pattern, text, re.IGNORECASE): This line finds all email addresses that match the email_pattern in the text. The results are stored in the email_addresses list.

for email in email_addresses:: This is a loop that iterates through the list of matched email addresses.

print(email): Within the loop, each matched email address is printed to the console.

The result is that the script searches the text for email addresses and extracts them, taking into account common patterns of email addresses while allowing for some variation in username and domain names. The re.IGNORECASE flag ensures that it works for both uppercase and lowercase email addresses.

# SETS
- A set is an unordered collections of unique elements, duplicates are not allowed.
- sets are used to store distinct values
-A set itself may be modified but the elements contained in the set must be hashable.
- sets are defiend by enclosing a coma-separated sequence of elements with curly brcaes '{}' or by using built in '**set()**' cinstructor.

### creating sets

```python
# using curly braces
my_set1 = {2,3}

# Using the set() constructor
my_set2 = set([1,2,4,4])

print(my_set1, my_set2)
```

### Uniqueness:

- Sets do not allow duplicate elements. If you try to add a duplicate element, it will not be stored in the set:

```python
# duplicates are removed in sets
my_set = {2,3,4,5,5,4}
print(my_set)
```

### No Indexing:

- Sets are unordered, and elements are not indexed. Therefore, you cannot access set elements by their position as you can with lists or tuples

### Common Set Operations:

- Sets support a variety of set operations, including union, intersection, difference, and more:

- **Union ('|')**: Combines two sets, returning elements that are in either set.
- **Intersection (&)**: Returns elements that are common to both sets.
- **Difference (-)** : Returns elements that are in the first set but not in the second.
- **Symmetric Difference (^)** : Returns elements that are in either of the sets, but not both.

## List Comprehensions
- List comprehensions is a concise and readable way to create lists. It offers a shorter syntax when you want to create a new list based on the values of an exisiting list.

### list comprehension syntax
- **new_list** = [expression *for* item *in* iterable *if* condition == *True*]
- **expression**: the expression to be evaluated for each item(element)
- **item(element)**: The variable representing each element in the iterable
- **iterable**: The iterable (e.g list, tuple, string ) to loop over
- **Condition(optional)**: An optional condition that filters which items are included in the new list. It only accepts items that evaluates to **True**

- the return value of a list comprehension is a new list that contains the modified elements that satisfy a given criteria.

### Example: Without list comprehension


```python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist = []

for x in fruits:
    if "a" in x:
        newlist.append(x)

print(newlist)
```

### Example: Using list comprehensions

```python
fruits = ["apple", "banana", "cherry", "kiwi", "mango", "passion"]
newlist = [x for x in fruits if "a" in x]

print(newlist)
```

### More Examples of list comprehensions use cases

#### Filtering odd numbers

```python
#using for loop
odd_numbers_for = []
for i in range (10):
    if i % 2 != 0:
        odd_numbers_for.append(i)

print(odd_numbers_for)        

```

```python
#using a list comprehension

odd_numbers_comp = [i for i in range(10) if i % 2 != 0]

print(odd_numbers_comp)
```

#### Creating Pairs from two Lists:


```python
# Using a for loop
pairs_for = []
for x in [1, 2, 3]:
    for y in ['a', 'b']:
        pairs_for.append((x, y))

print(pairs_for)
```

```python
#using for list comprehensions
pairs_comp = [(x,y) for x in [1,2,3,] for y in ['a', 'b']]

print(pairs_comp)
```

#### Nested List comprehensions


```python
matrix = []  # Initialize an empty list to store rows of the matrix

for i in range(3):
    # Append an empty sublist inside the list for each row
    matrix.append([])

    for j in range(5):
        # Append elements (in this case, numbers from 0 to 4) to the current row
        matrix[i].append(j)

print(matrix)
```

#### List cemprehension using if-else

#### List comprehension nested if statement


```python
# inserting numbers in the list which is a mutliple of 10-100
lis = [num for num in range(100) 
	if num % 5 == 0 if num % 10 == 0]
print(lis)
```

### Uses of a tuple
- retun multiple values from a function
- Can be used as keys in dictionaries as they are immutable
- Data integrity- used when data should remain constant.months.
- Variable swapping =e.g a,b = b,a

```python
#return multiple values from functions
def get_coordinates():
    x = 10
    y = 20
    z = 30
    return x, y, z

coordinates = get_coordinates()

print(coordinates) 
#coordinates is a tuple: (10, 20, 30)
```

### Example demonstrating a tuple

```python
# Complex Example: Employee Information

employee1 = ('John Doe', 30, 'Software Engineer', 75000.0)
employee2 = ('Alice Smith', 35, 'Data Scientist', 90000.0)

# Accessing information
#print(employee1[0])  # Output: 'John Doe'
#print(employee2[2])  # Output: 'Data Scientist'

# Concatenating tuples
team = (employee1, employee2)

# Returning multiple values from a function
def get_employee_details(employee):
    name, age, position, salary = employee
    return f"Name: {name}, Age: {age}, Position: {position}, Salary: ${salary}"

# Using the function to display details
for employee in team:
    print(get_employee_details(employee))
```
