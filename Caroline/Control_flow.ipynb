# Control Flow
- Control flow in Python refers to the order in which statements are executed in a script or program. Python provides several control flow constructs, including conditionals (if, elif, else), loops (for and while), and exception handling (try, except). 

### conditional statements ('if', 'elif', 'else'):
- used when you want to execute different blocks of code based on whether a certain condition is true or false.


````python
# if-elif statement example

letter = "A"

if letter == "B":
	print("letter is B")

elif letter == "C":
	print("letter is C")

elif letter == "A":
	print("letter is A")

else:
	print("letter isn't A, B or C")
````

````python
# Nested If Statement Example

# User authentication example
username = input("Enter your username: ")
password = input("Enter your password: ")

# Outer if statement
if username == "admin":
    # Nested if statement
    if password == "admin@123":
        print("Welcome, admin!")
    else:
        print("Invalid password for admin.")
else:
    print("Invalid username.")


````

### Loops: for loop and while loop
- Used when you want to repeatedly excute a block of code.

````python
# Using a for loop to iterate over a list
fruits = ["apple", "banana", "orange"]

for fruit in fruits:
    print(f"The fruit is: {fruit}")
````

**Looping through a dictionary**
- Through all key-value pairs
- through all keys
- through all Values

````python
#examples: looping through all key-value pairs
user_1 = {
    'username': 'carolkemmie',
    'first' : 'caroline',
    'last': 'kemmie'


}

for key, value in user_1.items():
    print(f"\nkey:{key}")
    print(f"\nvalue:{value}")
````

## **While loops**
- Used to repeatedly execute a block of code as long the specified condition is true. 
- Used when there is unknown number of iterations while *for* loop is used for a known number of iterations.
- The condition in a while loop continues to evaluates as long as the condition it true. So if the condition is always true, it will continue evaluating infinitely until the memory is full which can result to the program hannging.
- The condition continues to evaluate until the condition is false, that's when the loop terminates.

#### while loop syntax
```
while condition:
    # body of while loop
```

#### Examples using while loops

````python
#program displays numbers from 1 to 10

#initialize the variable
i = 1
n = 10

# while loop from i = 1 to 10
while i <= n :
    print(i)
    i += 1 #increament i
````

````python
# Python while loop with else
counter = 0

while counter < 5:
    print('Inside loop')
    counter = counter + 1
else:
    print('Inside else')
````

### use of break statement to terminate a while loop, 

````python
counter = 0

while counter < 3:
    # loop ends because of break
    # the else part is not executed 
    if counter == 1:
        break

    print('Inside loop')
    counter = counter + 1
else:
    print('Inside else')
````
